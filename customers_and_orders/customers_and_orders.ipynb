**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file
import csv

states = set()

with open(datafile, 'r') as fl:
    csvreader = csv.DictReader(fl)
    for row in csvreader:
        states.add(row['State'])

print(len(states)-1)
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
import csv

state_counts = {}

with open(datafile, 'r') as fl:
    csvreader = csv.DictReader(fl)
    for row in csvreader:
        state = row['State']
        if state in state_counts:
            state_counts[state] += 1
        else:
            state_counts[state] = 1


most_customers_state = max(state_counts, key=state_counts.get)
print(most_customers_state)
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file
import csv

state_counts = {}

with open(datafile, 'r') as fl:
    csvreader = csv.DictReader(fl)
    for row in csvreader:
        state = row['State']
        if state in state_counts:
            state_counts[state] += 1
        else:
            state_counts[state] = 1


least_customers_state = min(state_counts, key=state_counts.get)
print(least_customers_state, state_counts[least_customers_state])
```

**Question 5:** What is the most common last name?

```python
# open file and init vars
# evaluate the contents of the file
import csv

last_name_counts = {}

with open(datafile, 'r') as fl:
    csvreader = csv.DictReader(fl)
    for row in csvreader:
        last_name = row['LastName']
        if last_name in last_name_counts:
            last_name_counts[last_name] += 1
        else:
            last_name_counts[last_name] = 1

most_common_lastname = max(last_name_counts, key=last_name_counts.get)
print(most_common_lastname, last_name_counts[most_common_lastname])
```

```python
# open and evaluate orders.csv

import csv
from collections import defaultdict
from datetime import datetime

orders_file = "./sample_data/orders.csv"

orders = []


with open(orders_file, 'r') as fl:
    reader = csv.DictReader(fl)
    for row in reader:
        orders.append(row)

# ---------- Question 1: Unique Orders ----------
unique_orders = set(row['OrderID'] for row in orders)
print("Unique Orders:", len(unique_orders))

# ---------- Question 2 & 3: Average & Max items per order ----------
order_item_counts = defaultdict(int)
for row in orders:
    order_item_counts[row['OrderID']] += 1

average_items = sum(order_item_counts.values()) / len(order_item_counts)
max_items = max(order_item_counts.values())

print("Average number of items per order:", round(average_items, 2))
print("Highest number of items per order:", max_items)


from collections import defaultdict

# ---------- Question 5: Customer who spent the most in 2021 ----------
customer_spending = defaultdict(float)

for row in orders:
    date_str = row['Date']
    # filtramos todo 2021 usando startswith
    if date_str.startswith("2021-"):
        try:
            price = float(row['Price'])
            customer_spending[row['CustomerID']] += price
        except:
            continue  # ignora precios vacíos o mal formateados

if customer_spending:
    top_customer = max(customer_spending, key=customer_spending.get)
    print("Big Spender 2021:", top_customer, round(customer_spending[top_customer], 2))
else:
    print("No valid orders found for 2021.")

    from collections import defaultdict
import csv

orders_file = "./sample_data/orders.csv"

month_sales = defaultdict(float)

with open(orders_file, 'r') as fl:
    reader = csv.DictReader(fl)
    for row in reader:
        date_str = row['Date']
        # extraemos los primeros 7 caracteres: YYYY-MM
        month_key = date_str[:7]  # "2021-07", "2021-10", etc.
        try:
            month_sales[month_key] += float(row['Price'])
        except:
            continue  # ignora precios vacíos o mal formateados

# Obtenemos el mes con mayor venta
if month_sales:
    best_month = max(month_sales, key=month_sales.get)
    print("Best Month for Sales:", best_month, round(month_sales[best_month], 2))
else:
    print("No sales data found.")

    from collections import defaultdict
import csv

orders_file = "./sample_data/orders.csv"

month_sales = defaultdict(float)

with open(orders_file, 'r') as fl:
    reader = csv.DictReader(fl)
    for row in reader:
        date_str = row['Date']
        month_key = date_str[:7]  # "YYYY-MM"
        if month_key:  # ignoramos vacíos
            try:
                month_sales[month_key] += float(row['Price'])
            except:
                continue

# Obtenemos el mes con mayor venta
if month_sales:
    best_month_key = max(month_sales, key=month_sales.get)  # "YYYY-MM"
    # extraemos el número de mes
    best_month_number = int(best_month_key[5:7])  # "01" -> 1
    print("Best Month:", best_month_number)
else:
    print("No sales data found.")

import csv
from collections import defaultdict
import calendar

orders_file = "./sample_data/orders.csv"

month_sales = defaultdict(float)

with open(orders_file, 'r') as fl:
    reader = csv.DictReader(fl)
    for row in reader:
        date_str = row['Date']
        month_key = date_str[:7]  # "YYYY-MM"
        if month_key:
            try:
                month_sales[month_key] += float(row['Price'])
            except:
                continue

# Obtenemos el mes con más ventas
if month_sales:
    best_month_key = max(month_sales, key=month_sales.get)  # "YYYY-MM"
    best_month_number = int(best_month_key[5:7])  # "01" -> 1
    best_month_name = calendar.month_name[best_month_number]  # 1 -> "January"
    print("Best Month:", best_month_name)
else:
    print("No sales data found.")


```
